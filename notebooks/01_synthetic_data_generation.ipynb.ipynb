```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
np.random.seed(42)
NUM_SENSORS = 100
READINGS_PER_SENSOR = 1000
TOTAL_READINGS = NUM_SENSORS * READINGS_PER_SENSOR
sensor_ids = [f"S{i:03d}" for i in range(1, NUM_SENSORS + 1)]
end_date = datetime.now()
start_date = end_date - timedelta(days=730)
```

```python
reading_ids = np.arange(1, TOTAL_READINGS + 1)
sensor_for_reading = np.repeat(sensor_ids, READINGS_PER_SENSOR)
random_days = np.random.uniform(0, 730, TOTAL_READINGS)
reading_timestamps = [start_date + timedelta(days=float(d)) for d in random_days]
reading_values = np.round(np.random.normal(loc=50, scale=10, size=TOTAL_READINGS), 2)
data_sizes = np.random.randint(80, 201, TOTAL_READINGS)
```

```python
categories = np.array(["Hot"] * 20000 + ["Warm"] * 30000 + ["Cold"] * 30000 + ["Very Cold"] * 20000)
np.random.shuffle(categories)
```

```python
days_ago = np.zeros(TOTAL_READINGS, dtype=int)
hot_mask = categories == "Hot"
warm_mask = categories == "Warm"
cold_mask = categories == "Cold"
very_cold_mask = categories == "Very Cold"
days_ago[hot_mask] = np.random.randint(0, 15, hot_mask.sum())
days_ago[warm_mask] = np.random.randint(15, 61, warm_mask.sum())
days_ago[cold_mask] = np.random.randint(61, 181, cold_mask.sum())
days_ago[very_cold_mask] = np.random.randint(181, 731, very_cold_mask.sum())
last_accessed = [end_date - timedelta(days=int(d))for d in days_ago]
```

```python
sensor_df = pd.DataFrame({"READING_ID": reading_ids, "SENSOR_ID": sensor_for_reading, "READING_TIMESTAMP": reading_timestamps, "READING_VALUE": reading_values, "DATA_SIZE": data_sizes, "LAST_ACCESSED": last_accessed, "CATEGORY": categories})
```

```python
access_counts = np.zeros(TOTAL_READINGS, dtype=int)
access_counts[hot_mask] = np.random.randint(50, 201, hot_mask.sum())
access_counts[warm_mask] = np.random.randint(10, 51, warm_mask.sum())
access_counts[cold_mask] = np.random.randint(1, 11, cold_mask.sum())
access_counts[very_cold_mask] = np.random.randint(0, 2, very_cold_mask.sum())
total_accesses = access_counts.sum()
access_reading_ids = np.repeat(reading_ids,access_counts)
last_access_array = sensor_df["LAST_ACCESSED"].to_numpy(dtype="datetime64[ns]")
repeated_last_access = np.repeat(last_access_array,access_counts)
end_date_np = np.datetime64(end_date)
time_ranges = (end_date_np - repeated_last_access) / np.timedelta64(1, "s")
random_seconds = (np.random.random(total_accesses) * time_ranges)
access_timestamps = (repeated_last_access+ random_seconds.astype("timedelta64[s]"))
access_log_df = pd.DataFrame({"ACCESS_ID": np.arange(1, total_accesses + 1),"READING_ID": access_reading_ids,"ACCESS_TIMESTAMP": pd.to_datetime(access_timestamps)})
print("Total access events:", len(access_log_df))
display(access_log_df.head())
```

```python
oracle_sensor_df = sensor_df.drop(columns=["CATEGORY"])
```

```python
oracle_sensor_df.to_csv("sensor_readings.csv", index=False)
access_log_df.to_csv("data_access_log.csv", index=False)
```
